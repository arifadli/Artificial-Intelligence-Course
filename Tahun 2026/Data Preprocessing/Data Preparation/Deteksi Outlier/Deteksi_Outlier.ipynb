```python

import pandas as pd
import matplotlib.pyplot as plt

# 1. Membaca dataset
url = "https://raw.githubusercontent.com/arifadli/Paper-Penelitian/main/katalog_gempa.csv"
df = pd.read_csv(url)

# 2. Memilih data magnitudo yang valid
data = df["mag"].dropna()

# 3. Menghitung IQR
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1

# 4. Menentukan batas outlier
batas_bawah = Q1 - 1.5 * IQR
batas_atas = Q3 + 1.5 * IQR

# 5. Mengidentifikasi outlier
outlier = data[(data < batas_bawah) | (data > batas_atas)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Batas bawah:", batas_bawah)
print("Batas atas:", batas_atas)
print("Jumlah outlier:", len(outlier))

# 6. Visualisasi boxplot
plt.figure(figsize=(8, 4))
plt.boxplot(data, vert=False)
plt.xlabel("Magnitudo")
plt.title("Deteksi Outlier Magnitudo Gempa dengan IQR")
plt.grid(alpha=0.3)
plt.show()
```
