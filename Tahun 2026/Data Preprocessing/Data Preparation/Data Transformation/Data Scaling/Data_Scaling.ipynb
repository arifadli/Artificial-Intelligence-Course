```python

import pandas as pd
from sklearn.preprocessing import StandardScaler

# 1. Membaca dataset dari GitHub
url = "https://raw.githubusercontent.com/arifadli/Paper-Penelitian/main/katalog_gempa.csv"
df = pd.read_csv(url)

# 2. Memilih fitur numerik untuk scaling
fitur = ["mag", "depth", "lat", "lon"]
data = df[fitur].dropna()

# 3. Melakukan data scaling
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data)

# 4. Mengubah hasil scaling menjadi DataFrame
df_scaled = pd.DataFrame(
    data_scaled,
    columns=fitur,
    index=data.index
)

# 5. Menampilkan data sebelum dan sesudah scaling
print("Data sebelum scaling:")
print(data.head())

print("\nData setelah scaling:")
print(df_scaled.head())

# 6. Memeriksa hasil scaling
print("\nRata-rata setelah scaling:")
print(df_scaled.mean().round(3))

print("\nStandar deviasi setelah scaling:")
print(df_scaled.std(ddof=0).round(3))
```
